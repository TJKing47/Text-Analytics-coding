```python
# PART 1 — TF-IDF & Word2Vec with Preprocessing

import re
import numpy as np
from collections import Counter
from sklearn.cluster import KMeans
from sklearn.feature_extraction.text import TfidfVectorizer, ENGLISH_STOP_WORDS
from nltk.stem import PorterStemmer
from gensim.models import Word2Vec

dataset = [
    "I love playing football on the weekends",
    "I enjoy hiking and camping in the mountains",
    "I like to read books and watch movies",
    "I prefer playing video games over sports",
    "I love listening to music and going to concerts"
]

stemmer = PorterStemmer()

def preprocess(text):
    text = text.lower()
    text = re.sub(r'[^a-z\s]', ' ', text)
    words = text.split()
    words = [stemmer.stem(w) for w in words if w not in ENGLISH_STOP_WORDS]
    return " ".join(words)

def purity(y):
    total = len(y)
    count = Counter(y)
    return max(count.values()) / total
```

```python
# TF-IDF

vec1 = TfidfVectorizer()
X1 = vec1.fit_transform(dataset)

km1 = KMeans(n_clusters=2, random_state=42, n_init=10)
y1 = km1.fit_predict(X1)

print("TF-IDF without preprocessing Purity:", purity(y1))

clean_data = [preprocess(d) for d in dataset]

vec2 = TfidfVectorizer()
X2 = vec2.fit_transform(clean_data)

km2 = KMeans(n_clusters=2, random_state=42, n_init=10)
y2 = km2.fit_predict(X2)

print("TF-IDF with preprocessing Purity:", purity(y2))
```

```python
# Word2Vec

tokens1 = [d.split() for d in dataset]
w2v1 = Word2Vec(tokens1, vector_size=100, window=5, min_count=1, workers=4)

X_w1 = np.array([
    np.mean([w2v1.wv[w] for w in doc if w in w2v1.wv], axis=0)
    for doc in tokens1
])

km3 = KMeans(n_clusters=2, random_state=42, n_init=10)
y3 = km3.fit_predict(X_w1)

print("Word2Vec without preprocessing Purity:", purity(y3))

tokens2 = [d.split() for d in clean_data]
w2v2 = Word2Vec(tokens2, vector_size=100, window=5, min_count=1, workers=4)

X_w2 = np.array([
    np.mean([w2v2.wv[w] for w in doc if w in w2v2.wv], axis=0)
    for doc in tokens2
])

km4 = KMeans(n_clusters=2, random_state=42, n_init=10)
y4 = km4.fit_predict(X_w2)

print("Word2Vec with preprocessing Purity:", purity(y4))
```
