# Lab 3 - Web Scraping (Exercise Answers)

```python
import requests
from bs4 import BeautifulSoup
import csv

base_url = 'https://quotes.toscrape.com'
headers = {'User-Agent': 'Mozilla/5.0'}

einstein_quotes = []
life_quotes = []

def scrape_all_pages(url):
    while url:
        response = requests.get(url, headers=headers, timeout=10)
        print(response.status_code)

        soup = BeautifulSoup(response.text, 'html.parser')

        for quote in soup.find_all('div', class_='quote'):
            text = quote.find('span', class_='text').text
            author = quote.find('small', class_='author').text
            tags = ', '.join(tag.text for tag in quote.find_all('a', class_='tag'))

            if author == 'Albert Einstein':
                einstein_quotes.append({'Text': text, 'Author': author, 'Tags': tags})

            if 'life' in text.lower():
                life_quotes.append({'Text': text, 'Author': author, 'Tags': tags})

        next_page = soup.find('li', class_='next')
        url = base_url + next_page.find('a')['href'] if next_page else None

# Save Einstein quotes
with open('quotes_Einstein.csv', 'w', newline='', encoding='utf-8') as f:
    writer = csv.DictWriter(f, fieldnames=['Text', 'Author', 'Tags'])
    writer.writeheader()
    writer.writerows(einstein_quotes)

print('Einstein quotes saved!')

# Save life quotes
with open('life_quotes.csv', 'w', newline='', encoding='utf-8') as f:
    writer = csv.DictWriter(f, fieldnames=['Text', 'Author', 'Tags'])
    writer.writeheader()
    writer.writerows(life_quotes)

print('Life quotes saved!')
```
