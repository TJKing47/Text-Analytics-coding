# Lab 2 - Regular Expressions (Exercise Answers)

```python
import re

# 1. Extract all digits
text1 = "Order number: 24567, Tracking ID: 8934A12, Quantity: 30"
pattern1 = r"\d+"
print("Digits found:", re.findall(pattern1, text1))

# 2. Extract all non-alphanumeric characters
text2 = "Welcome! How's your day going? Call me @ 3pm."
pattern2 = r"[^a-zA-Z0-9]"
print("Non-alphanumeric characters:", re.findall(pattern2, text2))

# 3. Split text at commas and periods
text3 = "Data Science, AI, and Machine Learning. These fields are growing fast."
pattern3 = r"[,.]"
print("Split text:", re.split(pattern3, text3))

# 4. Replace email addresses
text4 = "Contact us at support@example.com or info@company.org"
pattern4 = r"\S+@\S+"
print("Anonymized text:", re.sub(pattern4, "[HIDDEN]", text4))

# 5. Extract domain types
text5 = "Emails: user@gmail.com, admin@university.edu, info@company.org"
pattern5 = r"@\w+\.(\w+)"
print("Domain types:", re.findall(pattern5, text5))
```
