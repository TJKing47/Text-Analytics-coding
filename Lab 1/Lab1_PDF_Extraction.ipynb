# Lab 1 - Working with PDF (Exercise)
Extract text from Business_Proposal.pdf
- Part 1: Extract ALL pages
- Part 2: Extract ONLY page 2


```python
import PyPDF2

# Read the PDF file
with open('Business_Proposal.pdf', 'rb') as file:
    reader = PyPDF2.PdfReader(file)
    
    # PART 1: Extract ALL pages
    all_text = ""
    for page in reader.pages:
        if page.extract_text():
            all_text += page.extract_text() + "\n"
    
    with open('business_proposal_all.txt', 'w', encoding='utf-8') as f:
        f.write(all_text)
    
    print("All pages extracted successfully!")

    # PART 2: Extract ONLY page 2 (index 1)
    page2_text = reader.pages[1].extract_text()
    
    with open('business_proposal_page_2.txt', 'w', encoding='utf-8') as f:
        f.write(page2_text)
    
    print("Page 2 extracted successfully!")
```
